```python
import numpy as np
import pandas as pd
import scipy .integrate as integrate
import scipy
import matplotlib.pyplot as plt
from scipy.special import gamma as gamma 
```

## 
$$
h(t; \nu) = \frac{\Gamma[(\nu + 1)/2]}{\Gamma(\nu/2)\sqrt{\pi\nu}}\left (1 + \frac{t^2}{\nu} \right)^{-(\nu + 1)/2}
$$

```python
def tdist (t, nu):
    return gamma((nu + 1)*0.5)/(gamma(nu*0.5)*np.sqrt(np.pi*nu))*(1.0 + t**2/nu)**(-(nu + 1)*0.5)
```

```python
nu = 14
x_array = np.linspace(-5, 5, 1000)
y_array = tdist(x_array, nu)
plt.plot(x_array, y_array, linestyle = '-', color = 'red', label = r'$\nu = %d$' %nu)
plt.legend(fontsize = 14)
plt.show()
```

```python
integrate.quad(tdist, -np.inf, np.inf, args = (nu))
```

## Ejercicio 1

```python
nu = 14 
t_array = np.linspace(-10, 10, 1000)

ss = 0
for tval in t_array:
    prob = integrate.quad(tdist, -np.inf, tval, args = (nu))
    if prob[0] >= 0.025:
        print(t_array[ss-1])
        break
    ss+=1
```

## Ejercicio de la clase.

```python
nu = 19 
t_array = np.linspace(-10, 10, 1000)

ss = 0
for tval in t_array:
    prob = integrate.quad(tdist, -np.inf, tval, args = (nu))
    if prob[0] >= 0.05:
        print(t_array[ss-1])
        break
    ss+=1
```
