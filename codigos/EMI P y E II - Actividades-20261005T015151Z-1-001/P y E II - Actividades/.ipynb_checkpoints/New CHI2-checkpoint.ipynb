```python
import numpy as np
import pandas as pd
import scipy .integrate as integrate
import scipy
import matplotlib.pyplot as plt
from scipy.special import gamma as gamma 
import scipy.stats as stats
```

```python
def tdist (t, nu):
    return gamma((nu + 1)*0.5)/(gamma(nu*0.5)*np.sqrt(np.pi*nu))*(1.0 + t**2/nu)**(-(nu + 1)*0.5)
```

```python
nu = 9
confianza = 0.95
alpha = 1 - confianza

# Para un intervalo bilateral al 95%, buscamos la probabilidad acumulada de 0.975 (1 - alpha/2)
t_valor = stats.t.ppf(1 - alpha / 2, df=nu)

print("El valor t crítico es:", t_valor)
```

```python
nu = 9

# Aumentamos el número de puntos para ganar precisión en los decimales
t_array = np.linspace(0, 10, 100000)

for tval in t_array:
    # Integración desde -inf hasta tval
    prob = integrate.quad(stats.t.pdf, -np.inf, tval, args=(nu))

    if prob[0] >= 0.975:
        # Imprimimos tval directamente (en lugar de t_array[ss-1])
        print("El valor t encontrado es:", tval)
        break
```
