```python
import numpy as np
from scipy.special import gamma as gamma
import scipy.integrate as integrate
import matplotlib.pyplot as plt
```

$$
h(t; \nu) = \frac{\Gamma[(\nu + 1)/2]}{\Gamma(\nu/2)\sqrt{\pi\nu}}\left(1 + \frac{t^2}{\nu} \right)^(-(\nu+1)/2)
$$

```python
def tdist(t, nu):
    return gamma((nu + 1)*0.5) / (np.sqrt(nu * np.pi) * gamma(nu * 0.5)) * (1 + t**2 / nu)**(-(nu + 1) * 0.5)
```

```python
nu = 14
x_arr = np.linspace(-5, 5, 1000)
y_arr = tdist(x_arr, nu)

plt.plot(x_arr, y_arr, linestyle="-", color="blue", label=r'$\nu = %d$' % nu)
```

```python
integrate.quad(tdist, -np.inf, np.inf, args=(nu))
```
